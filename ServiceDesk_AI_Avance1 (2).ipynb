**Objetivo:** Desarrollar un chatbot básico de soporte técnico utilizando Python y NLTK.

ServiceDesk AI está dirigido a usuarios que necesitan apoyo con problemas básicos de sus equipos de cómputo y cuentas de acceso.

La idea es facilitar una primera atención a las solicitudes más frecuentes, proporcionando respuestas sencillas sin que el usuario tenga que esperar inicialmente a un técnico.

Para esta primera versión utilizamos procesamiento de lenguaje natural y reglas, ya que estamos aprendiendo cómo identificar consultas y relacionarlas con respuestas previamente definidas.

```python

from nltk.tokenize import wordpunct_tokenize

mensaje = input("¿En qué puedo ayudarte? ")

palabras = wordpunct_tokenize(mensaje.lower())

if "bloqueada" in palabras or "bloqueado" in palabras:
    print("Intención: problema_bloqueo")
    print("¿Necesitas ayuda para desbloquear tu cuenta?")

elif "internet" in palabras or "red" in palabras:
    print("Intención: problema_internet")
    print("¿Otros equipos también tienen problemas de conexión?")


elif "contraseña" in palabras:
    print("Intención: restablecer_contrasena")
    print("¿Olvidaste tu contraseña y necesitas restablecerla?")


elif "lenta" in palabras or "lento" in palabras:
    print("Intención: equipo_lento")
    print("¿Tu computadora está lenta al iniciar o al utilizar aplicaciones?")

elif "no enciende" in mensaje.lower():
    print("Intención: equipo_no_enciende")
    print("¿Tu computadora no muestra ninguna señal de encendido?")



else:
    print("No identifico el problema")
```

## Arquitectura del chatbot

ServiceDesk AI es un chatbot básico de soporte técnico
desarrollado con Python y NLTK.

Su funcionamiento se divide en cuatro pasos:

1. **Usuario:** escribe el problema que necesita resolver.
2. **Procesamiento:** el mensaje se convierte a minúsculas
   y NLTK lo separa en palabras.
3. **Identificación de intención:** mediante reglas,
   el chatbot busca palabras clave relacionadas con
   problemas frecuentes.
4. **Respuesta:** muestra una respuesta previamente definida
   según el problema identificado.

Si el chatbot no reconoce la consulta, muestra el mensaje:
"No identifico el problema".

En esta primera versión no utilizamos aprendizaje automático.
Las respuestas se generan mediante reglas programadas.
